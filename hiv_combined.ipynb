```python
from rdkit import Chem
from rdkit.Chem import AllChem
import joblib
import numpy as np
```

```python
def predict_hiv_activity(smiles, model_path='models/xgb_hiv_2048bit.pkl', threshold=0.03918147):
    model = joblib.load(model_path)
    mol = Chem.MolFromSmiles(smiles)
    if mol is None:
        raise ValueError("Invalid SMILES string")
    fingerprint = AllChem.GetMorganFingerprintAsBitVect(mol, radius=2, nBits=2048)
    X = np.array(fingerprint).reshape(1, -1)
    probability = model.predict_proba(X)[0][1]
    prediction = (probability >= threshold).astype(int)
    return prediction, probability
```

```python
# make a smiles example
smiles_example = 'N=C(N)Nc1ccccc1SSc1ccccc1NC(=N)N.O=S(=O)(O)O' # this is active
smiles_example_2 = 'Nc1cccc(N=Nc2c(S(=O)(=O)O)cc3cc(NC(=O)Nc4ccc5c(O)c(N=Nc6cccc(N)c6)c(S(=O)(=O)O)cc5c4)ccc3c2O)c1' # this is active
smiles_example_3 = 'CCOC(=O)C1=CC=CC=C1C(=O)O' # this is inactive
smiles_example_4 = 'c1ccc(-c2nc3sc(-c4ccccc4)nc3s2)cc1' # this is inactive
aspiriin = 'CC(=O)OC1=CC=CC=C1C(=O)O'
```

```python
print('Aspirin prediction: class {}, probability {:.4f}'.format(*predict_hiv_activity(aspiriin)))
```

```python
print('Example molecule prediction: class {}, probability {:.4f}'.format(*predict_hiv_activity(smiles_example)))
```
